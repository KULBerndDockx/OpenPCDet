```python
import subprocess
def run_train(cfg_file, model, max_waiting_mins=1):
    process = subprocess.Popen(
        ['python3', 'train.py',
         '--cfg_file', cfg_file,
         '--pretrained_model', model,
         '--max_waiting_mins', str(max_waiting_mins)],
        stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1, cwd="/OpenPCDet/tools"
    )
    for line in process.stdout:
        print(line, end='')
    process.wait()
    return process.returncode
```
