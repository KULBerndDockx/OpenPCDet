```python
import subprocess
import os
def run_vis_kitti(cfg_file, ckpt, tag):
    process = subprocess.Popen(
        ['python3'    , 'demo.py',
         '--cfg_file' , cfg_file,
         '--ckpt'     , ckpt,
         'extra_tag'  , tag, 
         '--data_path', '/OpenPCDet/datasets/kitti/testing/velodyne',
        ],
        stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1, cwd="/OpenPCDet/tools"
    )
    for line in process.stdout:
        print(line, end='')
    process.wait()
    return process.returncode

def run_vis_nusc(cfg_file, ckpt, tag):
    process = subprocess.Popen(
        ['python3'    , 'demo.py',
         '--cfg_file' , cfg_file,
         '--ckpt'     , ckpt,
         'extra_tag'  , tag, 
         '--data_path', '/OpenPCDet/datasets/nuscenes/v1.0-mini/sweeps/LIDAR_TOP',
         '--ext'      , '.bin'
         ],
        stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1, cwd="/OpenPCDet/tools"
    )
    for line in process.stdout:
        print(line, end='')
    process.wait()
    return process.returncode

def run_vis_erod(cfg_file, ckpt, tag):
    process = subprocess.Popen(
        ['python3'    , 'demo.py',
         '--cfg_file' , cfg_file,
         '--ckpt'     , ckpt,
         'extra_tag'  , tag, 
         '--data_path', '/OpenPCDet/erod/points',
         '--ext'      , '.npy'],
        stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1, cwd="/OpenPCDet/tools"
    )
    for line in process.stdout:
        print(line, end='')
    process.wait()
    return process.returncode

```
