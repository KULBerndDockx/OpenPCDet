```python
import subprocess
def run_eval(cfg_file, ckpt_dir, max_waiting_mins=1):
    process = subprocess.Popen(
        ['python3', 'test.py',
         '--cfg_file', cfg_file,
         '--ckpt_dir', ckpt_dir,
         '--eval_all', '--max_waiting_mins', str(max_waiting_mins)],
        stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1
    )
    for line in process.stdout:
        print(line, end='')
    process.wait()
    return process.returncode
```
